## Descriptive Analysis.

### Descriptive analysis are used to understand and explore the values in the dataset and see what stands out. 

```python
# Import pandas
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("Data/CarPrice_Cleaned.csv")

df.head()
```

### The summary statistics shows the overview of the numerical columns, including avarage, minimum and maximum values.

```python
# Display summary statistics for the numerical columns in the dataset.
# This code is refferenced from LMS
df.describe()
```

### Findings

The dataset contains 205 cars with a wide range of prices. the avarage car price is around $13,277, with prices ranging from $5,118 to $45,400. There is also a noticable range in specifications such as engine size and horsepower, which can be explored further to see connection to the price. 

### Check Column Names.

The column names are checked to see which car details re available for the analysis.

```python
# Check the column names in the dataset
df.columns
```

### Avarage Price by Car Body Type.
The avarage price is compared accross the different car body types to see if there are noticable difference.
* AI Assistance: ChatGPT was used to help with wording and formulation in this next section.

```python
# Calculate the avarage price for each car body type
# Code created with ChatGPT assistance.
df.groupby('carbody')['price'].mean()
```

### Findings

Hardtop and convertible cars have the highest avarage prices, while hatchbacks have the lowest.  Conclusion is that avarage price differs noticably depending on the car body size

# Correlation Analysis.

Correlation is used to see how strongly the numerical car specifications are connected to price.
* AI Assistance: ChatGPT was used to help with wording and formulation in this next section.

```python
# Check corelations between engine Size and price.
# Code credited from LMS with small udjustments.
df['enginesize'].corr(df['price'])
```

### Findings

The corolation between engine size and price is approximately 0.87, showing a strong positive relationship. This suggests that cars with larger engines tend to have higher prices.

### Horsepower and Price Correlation.

The correlation between horsepower and price is checked to see how strongly they are related.

```python
# Check the correlation between horsepower and price.
# Code referenced from LMS with small adjustments.
df['horsepower'].corr(df['price'])
```

### Findings

The correlation between horsepower and price is approximately 0.81, showing  a strong  positive relationship. this suggests thta cars with higher horsepower tend to have higher prices.
