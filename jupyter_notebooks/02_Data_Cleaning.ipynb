## Data Cleaning

During thorough data inspection, a small inconsistency was found in the 'CarName' column. 
The same Audi model appears as both 'audi 100 ls' and 'audi 100ls'.
This will be corrected to so the car name is correct throughout the dataset.


```python
# Correct the misspelled car names in the CarName column.
# Code references from LMS
df['CarName']=df['CarName'].replace('audi 100ls', 'audi 100 ls')
```

```python
# Check the correct Audi car names after the correction.
df[df['CarName'].str.contains('audi',case=False)]['CarName']
```

The Audi car name was corrected and checked to make sure it is correct and consistent throughout the dataset.
